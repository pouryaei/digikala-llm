```python
from pathlib import Path
import sys

# اضافه کردن پوشه src به مسیر پایتون برای ایمپورت ماژول
root_dir = Path.cwd()
src_path = root_dir / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from src.digikala_ai.utils.normalization import normalize_persian_text

# تست تابع با نمونه متن‌های دیجی‌کالا
test_cases = [
    "گوشی موبایل سامسونگ مدل Galaxy A۵۴، ظرفیت ١٢٨ گیگابایت",
    "کتاب آموزش برنامه‌نویسی پایتون (جلدِ ۱)",
    "قيمت خريد كالا در تاریخ ١٤٠٣/٠٢/15"
]

for text in test_cases:
    cleaned = normalize_persian_text(text)
    print(f"RAW:   {text}")
    print(f"CLEAN: {cleaned}")
    print("-" * 50)
```
